```python
import numpy as np

a, b, n = 1, 3, 10
tau = (a + b) / 2  # true mean = 2
np.random.seed(0)
# Number of simulations (increase for more accuracy)
N = 200000
squared_errors = []

for _ in range(N):
    # 1) simulate a sample of size n from Uniform(a,b)
    x = np.random.uniform(a, b, size=n)
    # 2) compute MLEs of a and b: min and max
    a_hat = np.min(x)
    b_hat = np.max(x)
    # 3) MLE for tau = (a+b)/2 is (a_hat+b_hat)/2
    tau_hat = (a_hat + b_hat) / 2
    # 4) squared error
    squared_errors.append((tau_hat - tau)**2)

# Monte Carlo estimate of MSE
mse_hat = sum(squared_errors) / N
print("Estimated MSE of MLE tau_hat:", mse_hat)

# (Optional) plug-in estimator is sample mean; its MSE here equals Var(X)/n = ((b-a)^2/12)/n
mse_bar = ((b - a)**2 / 12) / n
print("MSE of sample mean (analytic):", mse_bar)
```
